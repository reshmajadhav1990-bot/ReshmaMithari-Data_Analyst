# TASK-1 : DATA CLEANING AND PRE-PROCESSING

## I have used CUSTOMER PERSONALITY ANALYSIS Dataset from Kaggle.com

### 1) Import The Necessary Library :

```python
import pandas as pd
```

### 2) Load the dataset from a CSV file

```python
df = pd.read_csv('marketing_campaign.csv', sep='\t')
print("Dataset loaded successfully.")
```

### 3) Display the first 5 rows and a summary of the data

```python
print("\nInitial Dataset Preview:")
print(df.head())
print("\nInformation summary:")
df.info()
df
```

### 4) Identify or Checking missing values

```python
# Checking for missing values

df.isnull().sum()   # isna() --- to find missing values or df.isna().sum()
```

### 5) Removal of missing rows

```python
#For columns with very few missing values, you can drop the entire row
df.dropna(inplace=True)

# Again Checking for missing values

df.isnull().sum()
```

### 6) Check and remove duplicate rows

```python
# Check for duplicate rows
print(f"\nNumber of duplicate rows: {df.duplicated().sum()}")
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
print("Duplicate rows have been removed.")
```

```python
# Verify that duplicates are gone
print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")
```

### 7) Rename column headers to be clean and uniform

```python
print("Original columns:")
print(df.columns)
```

```python
# Standardize text to a consistent case (e.g., lowercase)
# Clean the column headers in one step

df.columns = (df.columns
              .str.strip()               # Remove leading/trailing whitespace
              .str.lower()                # Convert all to lowercase
             )

print("\nCleaned columns:")
print(df.columns)
```

### 8) Check and fix data types

```python
# Check the data types

print(df.dtypes)
```

### Fix the data types

```python
# Convert 'Dt_Customer' to datetime Datatype
# Convert date formats to a consistent type

df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')

#Check the new data types
print(df.dtypes)
```

```python
# Convert 'Education' and'Marital_Status' to category Datatype

df['Education'] = df['Education'].astype('category')
df['Marital_Status'] = df['Marital_Status'].astype('category')

# Check the new data types
print(df.dtypes)
```

### 10) Removing outliers using the Interquartile Range (IQR)

```python
# Assuming 'Recency' is a column with potential outliers
Q1 = df['Recency'].quantile(0.25)
Q3 = df['Recency'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df['Recency'] >= lower_bound) & (df['Recency'] <= upper_bound)]
print("\nOutliers have been removed from 'Recency' using the IQR method.")
```

### 11) Review and Export the Cleaned Data

```python
print("\nFinal cleaned dataset information:")
df.info()

# Save the cleaned data to a new CSV file
df.to_csv('marketing_campaign_cleaned_dataset.csv', index=False)
print("\nCleaned dataset saved as 'marketing_campaign_cleaned_dataset.csv'.")
```
