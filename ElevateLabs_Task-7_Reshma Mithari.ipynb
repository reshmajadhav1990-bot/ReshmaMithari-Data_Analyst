# TASK-7 : GET BASIC SALES SUMMARY FROM A TINY SQLITE DATABASE USING PYTHON

## I have created sales_data.csv Dataset.

### 1) Import The Necessary Library :

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

### 2) Connect to the SQLite database :

```python
conn = sqlite3.connect('sales_data.db')
print("SQLite database connected successfully.")
```

### 3) Define and execute the SQL query :

```python
# Craft a SQL query to calculate the total quantity sold and total revenue per product. 

sales_summary_query = """
    SELECT
        product,
        SUM(quantity) AS total_quantity_sold,
        SUM(quantity * price) AS total_revenue
    FROM
        sales_data
    GROUP BY
        product
    ORDER BY
        total_revenue DESC;
    """
```

### 4) Load query results into a Pandas DataFrame :

```python
sales_summary_df = pd.read_sql_query(sales_summary_query, conn)
```

### 5) Display the sales summary :

```python
# Print the DataFrame to view the results.

print("Basic Sales Summary:")
print(sales_summary_df)
```

### 6)  Create a bar chart using Pandas' built-in plotting functionality :

```python
sales_summary_df.plot(kind='bar', x='product', y='total_revenue', title='Total Revenue by Product')
plt.xlabel("Product")
plt.ylabel("total_revenue")
plt.tight_layout() # Adjusts plot to prevent labels from overlapping
plt.show()
```

#### The top-performing product is the Smartphone with the highest revenue.

### 7) Save the chart as an image file :

```python
plt.savefig("sales_chart.png")
print("\nChart saved as sales_chart.png")
```

### 8) Close the database connection

```python
conn.close()
print("\nDatabase connection closed.")
```
