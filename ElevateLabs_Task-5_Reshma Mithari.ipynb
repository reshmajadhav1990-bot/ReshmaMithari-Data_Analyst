### 1) Import The Necessary Library :

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Ignore all warnings
warnings.filterwarnings("ignore")
```

### 2) Load the dataset from a CSV file

```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print("Datasets loaded successfully.")
```

### 3) Display the first 5 rows and a summary of the data

```python
print("\nInitial train.csv Dataset Preview:")
print(train.head())
print("\nInformation summary of train.csv:")
train.info()
train
```

```python
print("\nInitial test.csv Dataset Preview:")
print(test.head())
print("\nInformation summary of test.csv:")
test.info()
test
```

```python
print(train.shape)
print(test.shape)
```

### 4) Identify or Checking missing values

```python
# Checking for missing values for train.csv

train.isnull().sum()   # isna() --- to find missing values or df.isna().sum()
```

#### The 'Cabin' column contains numerous null values, so we will drop it.

```python
# Checking for missing values for train.csv

test.isnull().sum()   # isna() --- to find missing values or df.isna().sum()
```

#### The 'Cabin' column contains numerous null values, so we will drop it.

### 5) Handling Missing Values

```python
#For columns with very few missing values, you can drop the entire row
train.drop(columns=['Cabin'],inplace=True)
test.drop(columns=['Cabin'],inplace=True)

# Again Checking for missing values

train.isnull().sum()
```

```python
# Again Checking for missing values

test.isnull().sum()
```

### 6) Filling missing values (Imputation)

```python
# Filling missing values
train['Embarked'].fillna('S',inplace=True)

# Again Checking for missing values

train.isnull().sum()
```

```python
# Filling missing values
test['Fare'].fillna(test['Fare'].mean(), inplace=True)

# Again Checking for missing values

test.isnull().sum()
```

### 7) Creating a DataFrame for Easier Manipulation

```python
df=pd.concat([train,test],sort=True).reset_index(drop=True)
```

```python
print(df.shape)
```

```python
df.head()
```

#### The 'Age' column contains numerous null values, we need to fill in the null values in the 'Age' column with meaningful values.

```python
df.Age.fillna(df.Age.mean(),inplace=True)

# Again Checking for missing values

df.isnull().sum()
```

```python
df.head()
```

#### The 'Survived' column contains numerous null values, we need to fill in the null values in the 'Survived' column with mode.

```python
# Impute with the mode

mode_survived = df['Survived'].mode()[0]
df['Survived'].fillna(mode_survived, inplace=True)

# Again Checking for missing values
df.isnull().sum()
```

```python
df.head()
```

```python
df.tail()
```

### 8) Check and fix data types

```python
# Check the data types

print(df.dtypes)
```

### Fix the data types

```python
# Convert 'Age' and 'Survived' to category Datatype

df['Age'] = df['Age'].astype('int64')
df['Survived'] = df['Survived'].astype('int64')

# Check the new data types
print(df.dtypes)
```

### 9) Generate descriptive statistics

```python
df.describe()
```

### 10) Visualize and analyze relationships 

#### Univariate analysis: Examine individual features using visualizations.

```python
df.Survived.value_counts(normalize=True)

survival_ratio = df.Survived.value_counts(normalize=True)
print('Percentage of Survivors : ', round(survival_ratio[1]*100, 0), '%')
print('Percentage of Dead      : ', round(survival_ratio[0]*100, 0), '%')
```

```python
# Survival count
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.show()
```

##### In Survival Count plot, it's evident that the number of non-survivors is greater than survivors.
##### Only 26% of the passengers were survived, where as a majority 74% of the passenger did not survive the disaster.

```python
# Age distribution
sns.histplot(df['Age'], kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
# Passenger Class distribution
sns.countplot(x='Pclass', data=df)
plt.title('Passenger Class Distribution')
plt.show()
```

#### Bivariate analysis: Explore relationships between features and the target variable ('Survived').

```python
Survival_by_sex = df.groupby('Sex')[['Survived']].mean()

Survival_percentage_by_sex = Survival_by_sex * 100
print(Survival_percentage_by_sex)
```

```python
# Survival by Sex
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival Rate by Sex')
plt.show()
```

##### We can clearly see that male survial rates is around 13% where as female survial rate is about 50% which suggests that gender has a strong relationship with the survival rates.

```python
# Survival by Pclass

Survival_by_Pclass = df.groupby('Pclass')[['Survived']].mean()

Survival_percentage_by_Pclass = Survival_by_Pclass * 100
print(Survival_percentage_by_Pclass)
```

```python
# Survival by Pclass
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title('Survival Rate by Passenger Class')
plt.show()
```

##### There is also a clear relationship between Pclass and the survival by referring to above plot. Passengers on Pclass1 had a better survial rate of approx 42% whereas passengers on pclass3 had the worst survial rate of approx 17%

```python
# Survival by Age

Survival_by_Age = df.groupby('Age')[['Survived']].mean()

Survival_percentage_by_Age = Survival_by_Age * 100
print(Survival_percentage_by_Age)
```

```python
# Survival by Age
sns.histplot(x='Age', hue='Survived', data=df, kde=True)
plt.title('Survival by Age Distribution')
plt.show()
```

##### Children and younger individuals might have had a higher chance of survival. 

```python
# Survival by Fare
sns.histplot(x='Fare', hue='Survived', data=df, kde=True)
plt.title('Survival by Fare')
plt.show()
```

##### There is also a marginal relationship between the fare and survial rate.

```python
# Survival by Embarked Port
sns.barplot(x='Embarked', y='Survived', data=df)
plt.title('Survival Rate by Embarkation Port')
plt.show()
```
